```python
import random
import numpy as np
import torch
import torch.nn as nn
import torchvision.models as models
import torchvision

seed = 42

random.seed(seed)
np.random.seed(seed)
torch.manual_seed(seed)
torch.use_deterministic_algorithms(True, warn_only=False)
```

```python
batch_size = 32
dynamic_batch_size = True

model = models.vgg11_bn()
model.eval()
```

```python
in_tensor = torch.ones(batch_size, 3, 224, 224)

with torch.no_grad():
    out_tensor = model(in_tensor)
```

```python
out_tensor.shape, out_tensor[0][:100]
```

```python
for b in range(batch_size):
    for i in range(1000):
        print(f"{out_tensor[b][i]:.4f}", end=" ")
```

```python
params = {
    "model": model,
    "f": "vgg11.onnx",
    "args": in_tensor,
    "export_params": True,
    "opset_version": 18,
    "input_names": ["input"],
    "output_names": ["output"],
}

if dynamic_batch_size:
    batch_dim = torch.export.Dim("batch_size", min=1, max=1024)
    params["dynamic_shapes"] = ({0: batch_dim},)

_ = torch.onnx.export(**params)
```
